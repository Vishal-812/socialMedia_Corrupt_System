```python
df_2['timestamp'].unique()
```

```python
def convert_timestamp(t):
    t = str(t).strip()

    if t.isdigit():
        return pd.to_datetime(int(t), unit='s')

    return pd.to_datetime(t, dayfirst=True, errors='coerce')
```

```python
df_2['timestamp'] = df_2['timestamp'].apply(convert_timestamp)
```
